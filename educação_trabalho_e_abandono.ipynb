## 1. Introdução

A educação é um dos principais fatores relacionados às oportunidades de desenvolvimento social e profissional dos jovens. Neste projeto, investigamos o abandono escolar e indicadores do mercado de trabalho juvenil, buscando identificar padrões ao longo do tempo e diferenças entre as regiões brasileiras.

**Importante:** a análise de associação entre abandono escolar e mercado de trabalho não permite afirmar causalidade. Os resultados devem ser interpretados como evidências de padrões observados nas bases utilizadas.

## 2. Problema e objetivo

**Problema:** compreender como o abandono escolar se comporta no Brasil e como esse fenômeno se relaciona, em nível agregado, com indicadores de inserção dos jovens no mercado de trabalho.

**Objetivo:** analisar a evolução das taxas de abandono escolar e dos indicadores de desocupação e informalidade juvenil, comparando regiões e faixas etárias e verificando se existe associação entre esses indicadores.

## 3. Perguntas de análise

1. Como a taxa de abandono escolar evoluiu ao longo dos anos?
2. Quais regiões apresentam as maiores taxas de abandono escolar?
3. Existe relação entre o abandono escolar e a informalidade juvenil?
4. Existe relação entre o abandono escolar e a desocupação juvenil?
5. Como os indicadores de trabalho juvenil variam entre as faixas etárias de 14 a 17 anos e 18 a 24 anos?
6. Qual etapa de ensino apresenta maior taxa de abandono: Ensino Fundamental ou Ensino Médio?
7. Como a taxa de abandono escolar se comporta entre diferentes regiões e etapas de ensino?

## 4. Indicadores / KPIs

| KPI | Como é obtido | Justificativa |
|---|---|---|
| **Taxa de abandono — Ensino Fundamental** | Valor da variável `Taxa_Abandono_Total_EF` (%) | Mede a parcela de matrículas que deixou de frequentar o Ensino Fundamental e permite acompanhar a evolução do abandono nessa etapa. |
| **Taxa de abandono — Ensino Médio** | Valor da variável `Taxa_Abandono_Total_EM` (%) | Permite avaliar a permanência dos estudantes no Ensino Médio, etapa diretamente relacionada à transição para a vida profissional. |
| **Taxa de desocupação juvenil** | Indicador IBGE 4099 (%) | Mede a proporção de jovens na força de trabalho que está desocupada, permitindo acompanhar dificuldades de inserção profissional. |
| **Taxa de informalidade juvenil** | Indicador IBGE 12466 (%) | Permite avaliar a parcela dos ocupados inserida em condições informais de trabalho. |

Os KPIs serão calculados/selecionados diretamente a partir dos dados disponíveis, mantendo as definições das fontes oficiais.

### 5.2 IBGE — SIDRA, Tabela 4094

- **Instituição:** Instituto Brasileiro de Geografia e Estatística (IBGE)
- **Tabela:** 4094

- **Período disponível no arquivo:** 2013–2025
- **Formato:** CSV obtido a partir da API/JSON
- **Descrição:** indicadores do mercado de trabalho por grande região e grupo de idade. Para este trabalho, são utilizados principalmente os indicadores de taxa de desocupação (código 4099) e taxa de informalidade (código 12466).
- **Data de acesso:** 26/08/2026

### 5.3 Integração das fontes

As duas bases serão integradas pela combinação de **ano** e **Grande Região**. Para manter comparabilidade temporal, a análise integrada será limitada ao período comum entre as fontes, **2013–2023**. No INEP, serão utilizados registros com localização e dependência administrativa iguais a `Total`, evitando duplicação por diferentes recortes administrativos. No IBGE, serão consideradas as grandes regiões e os grupos etários disponíveis.

## 10. Tratamento dos dados

O tratamento remove apenas duplicidades exatas e padroniza as variáveis necessárias à análise. Valores ausentes não são substituídos automaticamente, pois a ausência pode representar uma informação não disponível na fonte.

```python
df_inep_tratado = df_inep.copy().drop_duplicates()
df_ibge_tratado = df_ibge.copy().drop_duplicates()

df_inep_tratado['Ano'] = pd.to_numeric(df_inep_tratado['Ano'], errors='coerce').astype('Int64')
for col in ['Taxa_Abandono_Total_EF', 'Taxa_Abandono_Total_EM']:
    df_inep_tratado[col] = pd.to_numeric(df_inep_tratado[col], errors='coerce')

df_ibge_tratado['V'] = pd.to_numeric(df_ibge_tratado['V'], errors='coerce')
df_ibge_tratado['D2C'] = pd.to_numeric(df_ibge_tratado['D2C'], errors='coerce').astype('Int64')
df_ibge_tratado['Ano'] = pd.to_numeric(df_ibge_tratado['D3C'].astype(str).str[:4], errors='coerce').astype('Int64')

print('Duplicatas INEP removidas:', len(df_inep) - len(df_inep_tratado))
print('Duplicatas IBGE removidas:', len(df_ibge) - len(df_ibge_tratado))
```

```python
print('Valores ausentes — INEP')
display(df_inep_tratado.isna().sum().sort_values(ascending=False).head(15))

print('Valores ausentes — IBGE')
display(df_ibge_tratado.isna().sum().sort_values(ascending=False).head(15))
```

## 11. Preparação do INEP para os KPIs

Para evitar que um mesmo ano/região apareça várias vezes por causa de recortes administrativos, utilizamos `Localizacao = Total` e `Dependencia_Administrativa = Total` quando esses valores existem. A unidade geográfica é usada como região de análise.

```python
inep_kpi = df_inep_tratado.copy()

if 'Localizacao' in inep_kpi.columns:
    inep_kpi = inep_kpi[inep_kpi['Localizacao'].astype(str).str.strip().str.lower().eq('total')]
if 'Dependencia_Administrativa' in inep_kpi.columns:
    inep_kpi = inep_kpi[inep_kpi['Dependencia_Administrativa'].astype(str).str.strip().str.lower().eq('total')]

inep_kpi = inep_kpi.dropna(subset=['Ano', 'Unidade_Geografica'])
inep_kpi = inep_kpi[['Ano', 'Unidade_Geografica', 'Taxa_Abandono_Total_EF', 'Taxa_Abandono_Total_EM']].copy()
inep_kpi = inep_kpi.rename(columns={'Unidade_Geografica':'Regiao'})

print('Regiões disponíveis no INEP:')
display(sorted(inep_kpi['Regiao'].dropna().unique().tolist()))
```

## 12. Análise exploratória — abandono escolar


```python
abandono_ano = inep_kpi.groupby('Ano', as_index=False)[['Taxa_Abandono_Total_EF','Taxa_Abandono_Total_EM']].mean()
display(abandono_ano.round(2))

plt.figure()
plt.plot(abandono_ano['Ano'], abandono_ano['Taxa_Abandono_Total_EF'], marker='o', label='Ensino Fundamental')
plt.plot(abandono_ano['Ano'], abandono_ano['Taxa_Abandono_Total_EM'], marker='o', label='Ensino Médio')
plt.title('Evolução da taxa média de abandono escolar — Brasil')
plt.xlabel('Ano')
plt.ylabel('Taxa de abandono (%)')
plt.legend()
plt.grid(alpha=0.25)
plt.show()
```

```python
abandono_regiao = inep_kpi.groupby('Regiao', as_index=False)[['Taxa_Abandono_Total_EF','Taxa_Abandono_Total_EM']].mean()
abandono_regiao = abandono_regiao.sort_values('Taxa_Abandono_Total_EM', ascending=False)
display(abandono_regiao.round(2))

plt.figure()
plt.bar(abandono_regiao['Regiao'].astype(str), abandono_regiao['Taxa_Abandono_Total_EM'])
plt.title('Taxa média de abandono no Ensino Médio por região')
plt.xlabel('Região')
plt.ylabel('Taxa de abandono (%)')
plt.xticks(rotation=30)
plt.show()
```

## 13. Análise exploratória — mercado de trabalho juvenil


```python
# Os códigos 4099 e 12466 correspondem, respectivamente, à taxa de desocupação e à taxa de informalidade.
ibge_indicadores = df_ibge_tratado[df_ibge_tratado['D2C'].isin([4099, 12466])].copy()
ibge_indicadores = ibge_indicadores.dropna(subset=['Ano','D1N','D4N','V'])
ibge_indicadores['Indicador'] = ibge_indicadores['D2C'].map({4099:'Taxa de desocupação', 12466:'Taxa de informalidade'})

display(ibge_indicadores[['Ano','D1N','D4N','Indicador','V']].head(10))
print('Indicadores encontrados:')
display(ibge_indicadores[['D2C','D2N']].drop_duplicates().sort_values('D2C'))
```

```python
mercado_ano = ibge_indicadores.groupby(['Ano','Indicador'], as_index=False)['V'].mean()
display(mercado_ano.pivot(index='Ano', columns='Indicador', values='V').round(2))

plt.figure()
for indicador, grupo in mercado_ano.groupby('Indicador'):
    plt.plot(grupo['Ano'], grupo['V'], marker='o', label=indicador)
plt.title('Evolução dos indicadores do mercado de trabalho juvenil')
plt.xlabel('Ano')
plt.ylabel('Taxa (%)')
plt.legend()
plt.grid(alpha=0.25)
plt.show()
```

```python
mercado_idade = ibge_indicadores.groupby(['D4N','Indicador'], as_index=False)['V'].mean()
display(mercado_idade.pivot(index='D4N', columns='Indicador', values='V').round(2))

plt.figure()
pivot_idade = mercado_idade.pivot(index='D4N', columns='Indicador', values='V')
pivot_idade.plot(kind='bar', ax=plt.gca())
plt.title('Indicadores do mercado de trabalho por faixa etária')
plt.xlabel('Faixa etária')
plt.ylabel('Taxa média (%)')
plt.xticks(rotation=0)
plt.show()
```

## 14. Integração INEP + IBGE

A integração utiliza **Ano + Região**. Como o INEP possui dados até 2023 no arquivo utilizado, o período integrado é 2013–2023. Para evitar misturar grupos etários diferentes, calculamos primeiro a média dos indicadores do IBGE entre as faixas etárias disponíveis e mantemos também uma análise separada por idade na seção anterior.

```python
ibge_regiao_ano = ibge_indicadores.groupby(['Ano','D1N','Indicador'], as_index=False)['V'].mean()
ibge_largo = ibge_regiao_ano.pivot_table(index=['Ano','D1N'], columns='Indicador', values='V').reset_index()
ibge_largo = ibge_largo.rename(columns={'D1N':'Regiao'})

integrado = inep_kpi.merge(ibge_largo, on=['Ano','Regiao'], how='inner')

print('Dimensão da base integrada:', integrado.shape)
display(integrado.head())
```

```python
correlacao = integrado[['Taxa_Abandono_Total_EF','Taxa_Abandono_Total_EM','Taxa de desocupação','Taxa de informalidade']].corr()
print('Correlação de Pearson entre os principais indicadores:')
display(correlacao.round(2))

fig, axes = plt.subplots(1, 2, figsize=(13, 5))
axes[0].scatter(integrado['Taxa de informalidade'], integrado['Taxa_Abandono_Total_EM'])
axes[0].set_title('Abandono no Ensino Médio × informalidade')
axes[0].set_xlabel('Taxa de informalidade (%)')
axes[0].set_ylabel('Abandono no Ensino Médio (%)')
axes[1].scatter(integrado['Taxa de desocupação'], integrado['Taxa_Abandono_Total_EM'])
axes[1].set_title('Abandono no Ensino Médio × desocupação')
axes[1].set_xlabel('Taxa de desocupação (%)')
axes[1].set_ylabel('Abandono no Ensino Médio (%)')
plt.tight_layout()
plt.show()
```
